## 1.实现hsv函数的相互转换
![](./rgb2hsv_formula20260928-1.png)


### 图中公式与 OpenCV 默认 HSV 转换实现的对比说明

图片中展示的是**标准数学定义下的 RGB 至 HSV 转换公式**。在实际工程落地中（特别是使用 OpenCV 时），核心数学逻辑保持一致，但存在以下关键差异：

---

#### 1. 数据范围与类型映射（最主要的工程差异）

* **图中的标准公式**：
  * **$V$ (Value)**：取值范围为 $[0, 255]$ 或 $[0, 1.0]$。
  * **$S$ (Saturation)**：取值范围为 $[0, 1.0]$。
  * **$H$ (Hue)**：取值范围为 $[0^\circ, 360^\circ)$。

* **OpenCV 默认实现（以 `uint8` / `CV_8U` 数据类型为例）**：
  OpenCV 为了将 HSV 结果存入标准的 8 位图像格式（数值范围 $0 \sim 255$），对数据范围进行了特殊缩放：
  * **$H$ 通道**：由于 $360^\circ$ 超出了 8 位无符号整数的最大值（$255$），OpenCV 将算出的色相角度**除以 2**，因此其取值范围映射为 **$[0, 179]$**。
  * **$S$ 通道**：将计算出的饱和度比例乘以 255，其取值范围映射为 **$[0, 255]$**。
  * **$V$ 通道**：直接保留最大通道值，取值范围为 **$[0, 255]$**。

> **注意**：若输入图像为 32 位浮点数类型（`CV_32F`，RGB 取值 $[0, 1.0]$），OpenCV 计算出的 $H$ 会保留为 $[0^\circ, 360^\circ)$，$S$ 与 $V$ 范围为 $[0, 1.0]$，此时数值范围与图中公式完全一致。

---

#### 2. 负角度处理（Hue 补正）

* **图中的标准公式**：
  当 $V = R$ 且 $G < B$ 时，计算公式 $\frac{60(G - B)}{V - \min(R,G,B)}$ 会输出**负角度**（例如 $-60^\circ \sim 0^\circ$）。
* **OpenCV 的处理**：
  OpenCV 要求色相必须落在正角度区间内。当计算结果为负数时，OpenCV 会自动加上 $360^\circ$（即进行 $H = H + 360^\circ$ 的修正），使其映射至 $[0^\circ, 360^\circ)$ 后再进一步做缩放。

---

#### 3. 颜色通道顺序（RGB vs BGR）

* **图中的标准公式**：默认输入通道顺序为 **RGB**。
* **OpenCV 默认实现**：OpenCV 默认的图像读取格式为 **BGR**。在调用 `cv2.cvtColor(img, cv2.COLOR_BGR2HSV)` 时，OpenCV 会自动按 BGR 通道对应关系进行推导。如果传入 RGB 格式的图像，需要使用 `cv2.COLOR_RGB2HSV` 转换标识。

# HSV 色彩空间转换到 RGB 色彩空间公式

给定 HSV 色彩空间的值：$H \in [0, 360)$, $S \in [0, 1]$, $V \in [0, 1]$（若 $V$ 范围为 $[0, 255]$，需先除以 $255$ 归一化）。

---

### 1. 中间变量计算

首先计算色差极差 $C$（Chroma）、主色调分量 $X$ 以及基准亮度 $m$：

* **极差 $C$**：
  $$C = V \times S$$

* **主色调分量 $X$**：
  $$H' = \frac{H}{60^\circ}$$
  $$X = C \times \left(1 - \vert{}(H' \bmod 2) - 1\vert{}\right)$$

* **基准亮度 $m$**：
  $$m = V - C$$

---

### 2. 临时 RGB 分量 $(R', G', B')$ 计算

根据色相角度 $H$ 所处的区间，确定 $(R', G', B')$ 的映射关系：

$$(R', G', B') = \begin{cases}  (C, X, 0), & 0^\circ \le H < 60^\circ \quad (0 \le H' < 1) \\ (X, C, 0), & 60^\circ \le H < 120^\circ \quad (1 \le H' < 2) \\ (0, C, X), & 120^\circ \le H < 180^\circ \quad (2 \le H' < 3) \\ (0, X, C), & 180^\circ \le H < 240^\circ \quad (3 \le H' < 4) \\ (X, 0, C), & 240^\circ \le H < 300^\circ \quad (4 \le H' < 5) \\ (C, 0, X), & 300^\circ \le H < 360^\circ \quad (5 \le H' < 6) \end{cases}$$

> **当 $S = 0$ 时**：为无彩色（灰度），此时直接有 $R' = G' = B' = 0$。

---

### 3. 最终 RGB 计算

加上基准亮度 $m$，得到最终归一化到 $[0, 1]$ 范围的 RGB 值：

$$R = R' + m$$
$$G = G' + m$$
$$B = B' + m$$

若需要映射回 **`0 ~ 255`** 的 8 位无符号整数区间（如 OpenCV 标准输出），直接乘以 $255$ 并取整即可：
$$R_{8u} = \text{round}(R \times 255)$$
$$G_{8u} = \text{round}(G \times 255)$$
$$B_{8u} = \text{round}(B \times 255)$$

```python
import numpy as np 
import matplotlib.pyplot as plt
import cv2 

def rgb2hsv(rgb_image):
    '''  in range rgb 0-255
        out range 
        h 0-180
        s 0-255
        v 0-255
    '''
    rows,cols,channels = rgb_image.shape
    rgb_01 = rgb_image.copy()
    rgb_01 = rgb_image.astype(np.float64)/255.0
    r = rgb_01[:,:,0]
    g = rgb_01[:,:,1]
    b = rgb_01[:,:,2]
    vmax =np.maximum(np.maximum(r,g) ,b)
    vmin = np.minimum(np.minimum(r,g),b)  
    H = np.zeros((rows,cols),dtype = np.float64)
    S = np.zeros((rows,cols),dtype = np.float64)
    V = np.zeros((rows,cols),dtype = np.float64)
    max_pos = np.argmax(rgb_01,axis=2)
    max_equal_min_pos = vmax==vmin # situation1 
    r_max = max_pos == 0
    g_bigger_b_pos = g >= b
    g_smaller_b_pos = g < b
    r_max_g_bigger_b = r_max & g_bigger_b_pos # situation 2
    r_max_g_smaller_b = r_max & g_smaller_b_pos # situation 3
    g_max = max_pos == 1 # situation 4
    b_max = max_pos == 2 #situation 5
    mask1 = max_equal_min_pos
    mask2 = r_max_g_bigger_b
    mask3 = r_max_g_smaller_b
    mask4 =  g_max 
    mask5 =  b_max
    H[mask1] = 0
    H[mask2] = 60 * ((g[mask2]-b[mask2])/(vmax[mask2]-vmin[mask2])) + 0
    H[mask3] = 60 * ((g[mask3]-b[mask3])/(vmax[mask3]-vmin[mask3])) + 360
    H[mask4] = 60 * ((b[mask4]-r[mask4])/(vmax[mask4]-vmin[mask4])) + 120
    H[mask5] = 60 * ((r[mask5]-g[mask5])/(vmax[mask5]-vmin[mask5])) + 240
    mask1_s = vmax==0
    mask2_s = vmax!=0
    S[mask1_s] = 0
    S[mask2_s] = 1 - vmin[mask2_s]/vmax[mask2_s]
    V = vmax
    H_new = H/2.0
    S_new = S*255
    V_new = V*255
    hsv_image = np.dstack((H_new,S_new,V_new))
    hsv_image_out = hsv_image.astype(np.uint8)
    return hsv_image_out

def hsv2rgb(hsv_image):

    '''  in range h 0-180
                  s 0-255
                  v 0-255
        out range rgb 0-255
    '''
    rows,cols,channels = hsv_image.shape
    H = hsv_image[:,:,0]
    S = hsv_image[:,:,1]
    V = hsv_image[:,:,2]

    H = (H.copy().astype(np.int32)*2)%360
    S = S.copy().astype(np.float64)/255.0
    V = V.copy().astype(np.float64)/255.0
    C= V * S
    m = V - C  
    X = C *(1- np.abs((H/60) % 2 -1 ))
    mask_a  = H >=0
    mask_b = H < 60
    mask0 = mask_a & mask_b
    mask_a = H>=60 
    mask_b = H < 120
    mask1 = mask_a & mask_b
    mask_a = H>=120 
    mask_b = H < 180
    mask2 = mask_a & mask_b 
    mask_a = H>=180 
    mask_b = H < 240
    mask3 = mask_a & mask_b
    mask_a = H>=240 
    mask_b = H < 300
    mask4 = mask_a & mask_b
    mask_a = H>=300 
    mask_b = H < 360
    mask5 = mask_a & mask_b
    Rp= np.zeros((rows,cols),dtype = np.float64)
    Gp = np.zeros((rows,cols),dtype = np.float64)
    Bp = np.zeros((rows,cols),dtype = np.float64)
    Rp[mask0] = C[mask0]
    Rp[mask1] = X[mask1]
    Rp[mask2] = 0
    Rp[mask3] = 0
    Rp[mask4] = X[mask4]
    Rp[mask5] = C[mask5]

    Gp[mask0] = X[mask0]
    Gp[mask1] = C[mask1]
    Gp[mask2] = C[mask2]
    Gp[mask3] = X[mask3]
    Gp[mask4] = 0
    Gp[mask5] = 0

    Bp[mask0] = 0
    Bp[mask1] = 0
    Bp[mask2] = X[mask2]
    Bp[mask3] = C[mask3]
    Bp[mask4] = C[mask4]
    Bp[mask5] = X[mask5]

    R = (Rp + m)*255
    G = (Gp + m)*255
    B = (Bp + m )*255
    rgb_image = np.dstack((R,G,B))
    rgb_image = rgb_image.astype(np.uint8)   
    return rgb_image



bgr_image = cv2.imread('onion.png')
b,g,r = cv2.split(bgr_image)
rgb_image = cv2.merge([r,g,b] )

hsv_image = rgb2hsv(rgb_image)
rgb_from_hsv_image = hsv2rgb(hsv_image)
hsv_opencv_image = cv2.cvtColor(rgb_image,cv2.COLOR_RGB2HSV)
rgb_from_hsv_opencv_image =cv2.cvtColor(hsv_opencv_image,cv2.COLOR_HSV2RGB)

plt.imshow(rgb_image)
plt.show()
plt.subplot(1,2,1)
plt.imshow(rgb2hsv(rgb_image))
plt.title('rgb2hsv')
plt.subplot(1,2,2)
plt.imshow(hsv_opencv_image)
plt.title('rgb2hsv opencv')
plt.show()



plt.subplot(1,2,1)
plt.imshow(rgb_from_hsv_image)
plt.title('hsv2rgb')
plt.subplot(1,2,2)
plt.imshow(rgb_from_hsv_opencv_image)
plt.title('hsv2rgb opencv')
plt.show()








```

# 11-3 YCbCr
$Y = 0.299*R + 0.587*G + 0.114*B$

$Cr = (R-Y)*0.713 + delta$

$Cb =(B-Y)*0.564 + delta$

$delta = 128 $

$M_{rgb2YCbCr} =
\begin{bmatrix}
  0.299 & 0.587 & 0.114 \\
  -0.169 & -0.331 & 0.500\\
  0.500 & -0.419 & -0.081
\end{bmatrix}
 \begin{bmatrix}
 R \\G\\B
 \end{bmatrix} +\begin{bmatrix} 0 \\128\\ 128\end{bmatrix}$

![](ycbcr.jpg)

## RGB 到 YCrCb 的矩阵表达式（BT.601）：

$$
\begin{bmatrix}
Y\\
Cr\\
Cb
\end{bmatrix}
=
\begin{bmatrix}
0.299 & 0.587 & 0.114\\
0.499813 & -0.418531 & -0.081282\\
-0.168736 & -0.331264 & 0.500000
\end{bmatrix}
\begin{bmatrix}
R\\
G\\
B
\end{bmatrix}
+
\begin{bmatrix}
0\\
128\\
128
\end{bmatrix}
$$

展开后：

$$
\begin{aligned}
Y &= 0.299R+0.587G+0.114B\\
Cr &= 0.499813R-0.418531G-0.081282B+128\\
Cb &= -0.168736R-0.331264G+0.5B+128
\end{aligned}
$$

其中：

- $R,G,B\in[0,255]$：输入 RGB 通道值。
- $Y$：亮度分量。
- $Cr$：红色差分量。
- $Cb$：蓝色差分量。
- $128$：色度通道的偏置。

### einsum 表示方法
$$ 
\mathrm{result}[i,j,s]
=
\sum_{k=0}^{2}
\mathrm{rgb}[i,j,k]\,
M_{\mathrm{ycrcb}}[s,k]
$$

```python
import numpy as np 
import matplotlib.pyplot as plt
import cv2 
def rgb2ycrcb(rgb):
    rgb=rgb.astype(np.float64)
    #M_ycrcb = np.array([[0.299,0.499813,-0.168935],[0.587,-0.418531,-0.331665],[0.114,-0.081282,0.50059]])
    M_ycrcb = np.array([
        [ 0.299,     0.587,     0.114   ],
        [ 0.499813, -0.418531, -0.081282],
        [-0.168736, -0.331264,  0.500000]
    ])
    ycrcb_bias = np.array([0,128,128])
    #result_numpy_multiply = rgb @ M_ycrcb.T + ycrcb_bias
    result_einsum = np.einsum('ijk,sk->ijs',rgb,M_ycrcb) + ycrcb_bias 
    result = np.clip(np.around(result_einsum,decimals = 0),0,255).astype(np.uint8)
    #result = np.clip(np.around(result_numpy_multiply,decimals = 0),0,255).astype(np.uint8)
    return result

bgr = cv2.imread('./leaves/IMG_20241104_074700.jpg')
bgr = cv2.resize(bgr,(1024,768))
b,g,r  = cv2.split(bgr)
rgb = cv2.merge([r,g,b])
ycrcb_result = rgb2ycrcb(rgb)

ycrcb_cv2 = cv2.cvtColor(rgb,cv2.COLOR_RGB2YCrCb)
fig,ax = plt.subplots(1,4,figsize=(15,5),dpi = 100)
ax[0].imshow(r,cmap = 'gray')
ax[0].set_title('red')
ax[1].imshow(g,cmap ='gray')
ax[1].set_title('green')
ax[2].imshow(b,cmap ='gray')
ax[2].set_title('blue')
ax[3].imshow(rgb)
ax[3].set_title('rgb')



ycrcb_cv2 = cv2.cvtColor(rgb,cv2.COLOR_RGB2YCrCb)
y,cr,cb = cv2.split(ycrcb_result)
fig,ax = plt.subplots(1,4,figsize=(15,5),dpi = 100)
ax[0].imshow(y,cmap = 'gray')
ax[0].set_title('Y')
ax[1].imshow(cr,cmap ='gray')
ax[1].set_title('cr')
ax[2].imshow(cb,cmap ='gray')
ax[2].set_title('cb')
ax[3].imshow(ycrcb_result)
ax[3].set_title('ycrcb_result')


print(np.allclose(ycrcb_result,ycrcb_cv2,atol=1))
print(ycrcb_result[0:2,0:2,:])
print(ycrcb_cv2[0:2,0:2,:])
```

# 11-4 YIQ 模型


$M_{RGB2YIQ} =\begin{bmatrix}
0.299 &0.587& 0.114 \\
0.596 &  -0.274 & -0.322\\ 
0.211 &-0.523 &0.312
\end{bmatrix}
$
注意M[2,1],教材上错误印刷为-0.253

```python
import numpy as np 
import matplotlib.pyplot as plt
import cv2 
def rgb2yiq(rgb):
    rgb=rgb.astype(np.float64)
    yiq = np.array([[0.299,0.587,0.114],[0.596,-0.274,-0.322],[0.211,-0.523,0.312]]).T    
    result_einsum = np.einsum('ijk,sk->ijs',rgb,yiq)   # 特别需要注意这里是sk，而不是ks
    result = np.clip(np.around(result_einsum,decimals = 0),0,255).astype(np.uint8)
    return result

bgr = cv2.imread('./leaves/IMG_20241104_074700.jpg')
bgr = cv2.resize(bgr,(1024,768))
b,g,r  = cv2.split(bgr)
rgb = cv2.merge([r,g,b])
yiq_result = rgb2yiq(rgb)

fig,ax = plt.subplots(1,4,figsize=(15,5),dpi = 100)
ax[0].imshow(r,cmap = 'gray')
ax[0].set_title('red')
ax[1].imshow(g,cmap ='gray')
ax[1].set_title('green')
ax[2].imshow(b,cmap ='gray')
ax[2].set_title('blue')
ax[3].imshow(rgb)
ax[3].set_title('rgb')

y,i,q = cv2.split(yiq_result)
fig,ax = plt.subplots(1,4,figsize=(15,5),dpi = 100)
ax[0].imshow(y,cmap = 'gray')
ax[0].set_title('Y')
ax[1].imshow(i,cmap ='gray')
ax[1].set_title('i')
ax[2].imshow(q,cmap ='gray')
ax[2].set_title('q')
ax[3].imshow(yiq_result)
ax[3].set_title('ycrcb_result')

```
